```
dn = int(input("¿Qué día naciste?: "))
mn = int(input("¿Qué mes naciste?: "))
an = int(input("¿Qué año naciste?: "))

da = int(input("Día actual: "))
ma = int(input("Mes actual: "))
aa = int(input("Año actual: "))

edad = aa - an

# Condición para saber si ya cumplió años
if ma < mn or (ma == mn and da < dn):
    edad = edad - 1

print("edad:",edad)
```
