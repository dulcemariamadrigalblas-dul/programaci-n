```
a=int(input("¿cuántas calificaciones desea capturar? ingrese el número de materias: "))
suma=0

for i in range(1,a+1):
  calif=int(input("ingresa la calificación de una a una: "))
  suma=suma+calif


promedio=suma/a
print("el promedio es",promedio )




```
